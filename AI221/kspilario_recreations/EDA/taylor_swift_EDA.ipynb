# Exploratory Data Analysis: Taylor Swift Spotify Data Set

Reference: https://www.kaggle.com/code/aaronjones32/predicting-song-popularity-from-spotify-dataset

```python
# imports
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# reading and displaying Taylor Swift DataFrame
ts = pd.read_csv("taylor_swift_spotify.csv")
display(ts)
```

```python
# displaying preliminary information about the dataset
ts.info()
```

## Correlation map for numerical features

```python
# initializing the figure
plt.figure(figsize=(12, 6))
#sns.set(style="whitegrid")

# getting the correlation among each variable, except the "Unnamed: 0" and "track_number columns
corr = ts.drop(columns=["Unnamed: 0", "name", "album", "release_date", "track_number", "id", "uri"]).corr()

# visualizing the correlation as a heatmap
sns.heatmap(corr, annot=True, cmap="viridis")
plt.xticks(rotation=30)
plt.show()
```

```python
# checking how many songs per album
ts['album'].value_counts()
```

## Visualize album characteristics as box plots

```python
# collect all features having [0, 1] as the range of values
melted_ts = pd.melt(ts, id_vars="album", value_vars=ts.columns.values[np.r_[7:12, 13, 15]]) # selecting only columns 7:11, 13, and 15 from the ts DataFrame
melted_ts.columns = ['Album', 'Features', 'Values'] # renaming the columns of the melted DataFrame
display(melted_ts)
```

```python
# creating boxplots per album
plt.figure(figsize=(10,20))
sns.boxplot(data=melted_ts, width=0.8, orient="h", x="Values", y="Features", hue='Album', palette="colorblind") # width parameter to control the width of the box
#plt.rc('axes', labelsize=15) # setting the axes font size to 15 for the succeeding plots
plt.legend(bbox_to_anchor=(1, -0.05))
plt.grid()
plt.show()
```

```python
# creating boxplots for the duration of each song per album
plt.figure(figsize=(10,15))
ts["duration_min"] = ts["duration_ms"] / 60000 # converting duration in ms to min
sns.boxplot(data=ts, x="duration_min", y="album", hue="album", palette="colorblind")
plt.grid()
plt.show()

ts.drop(columns="duration_min", inplace=True) # discarding the newly-created duration_min column
```

```python
# checking the distribution of popularity per album
plt.figure(figsize=(10,15))
sns.boxplot(data=ts, x="popularity", y="album", hue="album", palette="colorblind")
plt.grid()
plt.show()
```

## Rankings from the data set

```python
# ranking songs by their property from acousticness column (7th column) up to the duration_ms (last column)
# looping through columns 7 up to the end
for prop in ts.columns.values[7:]:
    print(f'Ranking by {prop}')
    sorted_ts = ts.sort_values(by=prop, ascending=False)[["name", prop]] # soring values by song name and property
    display(sorted_ts.drop_duplicates(subset='name')) # displaying the sorted DataFrame, dropping duplicates by name
    print("====================\n")
```

## Visualizing the frequency of release dates

```python
ts['release_date'] = pd.to_datetime(ts['release_date']) # converting release_date column to datetime type to easily extract month, day, and year
month_of_release = ts['release_date'].dt.month # extracting the month
day_of_month = ts['release_date'].dt.day # extracting the day
year_of_release = ts['release_date'].dt.year # extracting the year

# initializing the whole plot to have 3 columns of subplots
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(20,4), sharey=False)

# creating countplots of songs per day, month, and year of release
sns.countplot(x=day_of_month, ax=axes[0], hue=day_of_month, legend=False, palette="colorblind")
sns.countplot(x=month_of_release, ax=axes[1], hue=month_of_release, legend=False, palette="colorblind")
sns.countplot(x=year_of_release, ax=axes[2], hue=year_of_release, legend=False, palette="colorblind")

# setting the titles for each subplot
axes[0].set_title('Day of month released', fontsize=15)
axes[1].set_title('Month of release', fontsize=15)
axes[2].set_title('Year of release', fontsize=15)

plt.show()
```

## Visualizing evolutions of data with time

```python
ts["release_year"] = year_of_release

# checking evolution of the properties per time
# looping through columns 7 to 15 (acousticness to valence)
for prop in ts.columns.values[7:16]:
    means = ts.groupby("release_year")[prop].mean() # getting the mean of each property column by year
    plt.figure(figsize=(8,3))
    plt.plot(means.index, means.loc[:], 'bo-', linewidth=2)
    plt.xticks(np.arange(2006,2024), rotation=30)
    plt.ylabel(f"Average {prop}")
    plt.xlabel('Year')
    plt.title(f"Average {prop} through the years", fontsize=15)
    plt.show()
    print()
```
